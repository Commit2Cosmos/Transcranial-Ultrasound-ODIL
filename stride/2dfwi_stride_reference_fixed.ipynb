# Stride reference solutions matched to odil_wave Shepp–Logan setup

This notebook produces Stride reference solutions aligned with the odil_wave
experiment config (250×250 interior, scale=0.85, ring frac=0.9, t_max=500 µs,
80 kHz tone burst, 100 transducers, continuation 40–100 kHz):

1. Acoustic forward modelling for synthetic observations.
2. Stride FWI from homogeneous water (ODIL `init.profile=homogeneous`).
3. Stride FWI from an exact-skull / water-interior start.
4. Stride FWI from a smooth-skull start (ODIL `shepp_logan_skull` with α, σ).

Physics/acquisition knobs match ODIL where possible. Solver differences remain
(time-domain FWI + GradientDescent vs Helmholtz ODIL + LBFGS-B).

Running this notebook in the stride conda environment.

## 4. Configure the coarse-grid Stride FWI operator

Both starts use the same observations, inverse grid, loss, frequency schedule,
shot selection and physical bounds.

```python
max_freqs = list(FREQ_BANDS)

# PPW check for each FWI stage (using minimum velocity = water)
print(f"{'Stage':<6} {'f_max':>10} {'PPW':>8}")
for i, f in enumerate(max_freqs, 1):
    ppw = ref.SOS_WATER / (f * meta_inv["spacing"][0])
    print(f"{i:<6} {f/1e3:>8.0f} kHz  {ppw:>5.1f}")
```

```python
pde_inv = IsoAcousticDevito.remote(
    grid=problem_inv.grid,
    len=runtime.num_workers,
)

# Match ODIL continuation bands; use all shots (ODIL joint schedule over 100 sources).
max_freqs_water = list(FREQ_BANDS)
max_freqs_skull = list(FREQ_BANDS)
max_freqs_smooth = list(FREQ_BANDS)
num_iters = 15
shots_per_iteration = N_RECEIVERS

def make_fwi_components(vp):
    loss = L2DistanceLoss.remote(len=runtime.num_workers)
    optimiser = GradientDescent(
        vp,
        step_size=10,
        process_grad=ProcessGlobalGradient(),
        process_model=ProcessModelIteration(min=C_CLAMP[0], max=C_CLAMP[1]),
    )
    return loss, optimiser, OptimisationLoop()

async def run_fwi(vp, frequencies, out_dir, *, run_id="stride_fwi"):
    """FWI with ODIL-style per-band metrics, snapshots, and c_bands.gif."""
    ref.set_medium_vp(problem_inv, vp)
    ref.check_forward_ready(problem_inv)
    loss, optimiser, optimisation_loop = make_fwi_components(vp)

    recorder = ref.BandRunRecorder(
        out_dir,
        vp_true_inv_array,
        phantom_scale=SHEPP_SCALE,
        run_id=run_id,
    )
    # Initial model (before any band)
    row0 = recorder.record(band_index=0, frequency_hz=None, vp=vp, label="initial")
    print(
        f"[initial] rms={row0['rms']:.1f}  rel_c={row0['rel_c_error']:.4f}  "
        f"ssim_head={row0['ssim_head_roi']}"
    )

    rms_history = [row0["rms"]]
    rel_history = [row0["rel_c_error"]]
    ssim_history = [row0["ssim_head_roi"]]
    band_labels = ["Initial"]

    for band_i, (_, frequency) in enumerate(optimisation_loop.blocks(len(frequencies), frequencies)):
        await adjoint(
            problem_inv,
            pde_inv,
            loss,
            optimisation_loop,
            optimiser,
            vp,
            num_iters=num_iters,
            select_shots=dict(
                num=shots_per_iteration,
                every=1,
                randomly=False,
            ),
            f_max=frequency,
            max_freqs=frequencies,
            save_undersampling=8,
            kernel='OT4',
            fw3d_mode=True,
            interpolation_type='hicks',
        )
        row = recorder.record(
            band_index=band_i + 1,
            frequency_hz=float(frequency),
            vp=vp,
        )
        print(
            f"[band {band_i + 1} {frequency/1e3:.0f} kHz] "
            f"rms={row['rms']:.1f}  rel_c={row['rel_c_error']:.4f}  "
            f"ssim_head={row['ssim_head_roi']}"
        )
        rms_history.append(row["rms"])
        rel_history.append(row["rel_c_error"])
        ssim_history.append(row["ssim_head_roi"])
        band_labels.append(f"{frequency/1e3:.0f} kHz")

    recorder.save_final(vp)
    gif_path = recorder.write_gif(
        domain_m=DOMAIN_M,
        filename="c_bands.gif",
        title=f"Recovered vp — {run_id}",
    )
    print("Wrote", recorder.jsonl_path)
    print("Wrote", gif_path)
    return vp, {
        "rms": rms_history,
        "rel_c_error": rel_history,
        "ssim_head_roi": ssim_history,
        "labels": band_labels,
        "recorder": recorder,
        "gif": gif_path,
    }
```

## 5. FWI from homogeneous water

The inversion parameter belongs to problem_inv.grid, not the fine reference
grid.

```python
water_start_array = ref.water_model(INV_INTERIOR)

vp_water_start = ref.inversion_vp_from_array(
    problem_inv,
    water_start_array,
)

# Plot the initial homogeneous-water model
ref.plot_vp(
    water_start_array,
    domain_m=DOMAIN_M,
    title="Initial model: homogeneous water",
    vmin=1400,
    vmax=3000
)

# Run FWI
water_dir = OUT / "fwi" / "water_start"
vp_water_recovered, _hist_water = await run_fwi(
    vp_water_start, max_freqs_water, water_dir, run_id="water_start"
)
_rms_water = _hist_water["rms"]
_labels_water = _hist_water["labels"]
ref.save_vp_npy(vp_water_recovered, water_dir / "vp_recovered.npy")

# Plot the recovered model
ref.plot_vp(
    vp_water_recovered,
    domain_m=DOMAIN_M,
    title="Recovered: water initialization",
    vmin=1400,
    vmax=3000
)

print("Water-start FWI saved to", water_dir)
```

```python
# Convergence: water start
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(range(len(_rms_water)), _rms_water, "o-", color="steelblue", label="Water start")
ax.set_xticks(range(len(_labels_water)))
ax.set_xticklabels(_labels_water)
ax.set_xlabel("Optimisation stage")
ax.set_ylabel("RMS error vs. true model  (m/s)")
ax.set_title("FWI convergence, water initialisation")
ax.grid(alpha=0.3)
ax.legend()
plt.tight_layout()
plt.show()
print(f"Water-start RMS: {_rms_water[0]:.1f} -> {_rms_water[-1]:.1f} m/s")
print("GIF:", _hist_water["gif"])
```

## 7. Compare and export

The FWI outputs are already on the intended inversion grid. No post-hoc
downsampling is used to pretend that a fine-grid inversion was coarse.

```python
# Grid-convergence compare from saved observed traces.
#
# Fine:  OUT/forward_fine/shot_XXXX_observed.npy  (+ grid_meta_fine.json)
# Inv:   OUT/forward_inv_independent/...          (+ grid_meta_inversion.json)
#
# Set RECOMPUTE_INV_FORWARD=True once (needs mosaic + problem_inv + truth) to
# generate the independent inv-grid forward if those files are missing.

RECOMPUTE_INV_FORWARD = False
check_shot_gc = 30
check_rx_gc = 0

fine_dir = OUT / "forward_fine"
inv_dir = OUT / "forward_inv_independent"

if RECOMPUTE_INV_FORWARD or not (inv_dir / f"shot_{check_shot_gc:04d}_observed.npy").is_file():
    if "problem_inv" not in globals() or "vp_true_inv_field" not in globals():
        raise RuntimeError(
            "Independent inv forward not on disk and problem_inv/vp_true_inv_field "
            "are unavailable. Re-run problem-build + fine forward, then set "
            "RECOMPUTE_INV_FORWARD=True."
        )
    # Clear transferred observations so the forward solver actually runs
    for shot in problem_inv.acquisitions.shots:
        if shot.observed is not None and shot.observed.allocated:
            shot.observed.deallocate()

    pde_inv_gc = IsoAcousticDevito.remote(
        grid=problem_inv.grid, len=runtime.num_workers
    )
    await forward(
        problem_inv,
        pde_inv_gc,
        vp_true_inv_field,
        dump=False,
    )
    ref.export_shot_observed_npy(problem_inv, inv_dir)
    print(f"Saved independent inv forward -> {inv_dir}")

    # Restore transferred observations for FWI if the fine problem is still live
    if "problem_ref" in globals():
        ref.transfer_observed_to_problem(
            problem_ref, problem_inv, check_geometry=True
        )
        print("Transferred observations restored — FWI data intact")

# --- plot from disk ---
fine_meta = ref.load_grid_meta(OUT / "grid_meta_fine.json")
inv_meta = ref.load_grid_meta(OUT / "grid_meta_inversion.json")
fine_shape = tuple(fine_meta["interior_shape"])
inv_shape = tuple(inv_meta["interior_shape"])

fine_t_gc = ref.time_axis_from_meta(fine_meta) * 1e6  # µs
inv_t_gc = ref.time_axis_from_meta(inv_meta) * 1e6

fine_all_gc = ref.load_observed_shot(fine_dir, check_shot_gc)
inv_all_gc = ref.load_observed_shot(inv_dir, check_shot_gc)
fine_rx_gc = fine_all_gc[check_rx_gc]
inv_rx_gc = inv_all_gc[check_rx_gc]

fig, ax = plt.subplots(figsize=(10, 3))
ax.plot(fine_t_gc, fine_rx_gc, lw=1, label=f"Fine {fine_shape[0]}x{fine_shape[1]} (reference)")
ax.plot(
    inv_t_gc,
    inv_rx_gc,
    lw=1.5,
    ls="--",
    label=f"Inv {inv_shape[0]}x{inv_shape[1]} (independent forward)",
)
ax.set_xlabel("time (µs)")
ax.set_ylabel("pressure")
ax.set_title(
    f"Grid convergence — receiver {check_rx_gc}, shot {check_shot_gc}: independent forwards"
)
ax.legend(fontsize=8)
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()

_clip_gc = float(np.percentile(np.abs(fine_all_gc), 98))
_norm_gc = plt.matplotlib.colors.TwoSlopeNorm(vmin=-_clip_gc, vcenter=0.0, vmax=_clip_gc)

fig, axes = plt.subplots(1, 2, figsize=(13, 4), sharey=True)
for ax, data, t_ax, label in [
    (axes[0], fine_all_gc, fine_t_gc, f"Fine {fine_shape[0]}x{fine_shape[1]}"),
    (axes[1], inv_all_gc, inv_t_gc, f"Inv {inv_shape[0]}x{inv_shape[1]} (independent)"),
]:
    im = ax.imshow(
        data,
        aspect="auto",
        origin="upper",
        cmap="RdBu_r",
        norm=_norm_gc,
        extent=[t_ax[0], t_ax[-1], data.shape[0], 0],
    )
    ax.set_title(label)
    ax.set_xlabel("time (µs)")
    ax.set_ylabel("receiver")
fig.suptitle("Independent forwards on each grid", y=1.02)
plt.tight_layout()
fig.subplots_adjust(right=0.88)
cbar_ax = fig.add_axes([0.91, 0.15, 0.015, 0.7])
fig.colorbar(im, cax=cbar_ax, label="pressure")
plt.show()

fine_peak_t = fine_t_gc[np.argmax(np.abs(fine_rx_gc))]
inv_peak_t = inv_t_gc[np.argmax(np.abs(inv_rx_gc))]
amp_ratio = float(np.max(np.abs(inv_rx_gc))) / float(np.max(np.abs(fine_rx_gc)) + 1e-30)

print(f"\nGrid convergence statistics (receiver {check_rx_gc}, shot {check_shot_gc})")
print(f"  Fine peak : {np.max(np.abs(fine_rx_gc)):.4e}  at {fine_peak_t:.3f} µs")
print(f"  Inv  peak : {np.max(np.abs(inv_rx_gc)):.4e}  at {inv_peak_t:.3f} µs")
print(f"  Amplitude ratio (inv/fine) : {amp_ratio:.4f}")
print(f"  Peak-time difference       : {abs(fine_peak_t - inv_peak_t):.4f} µs")
print(
    f"  Fine dt = {fine_meta['time_step_s']*1e6:.4f} µs,  "
    f"Inv dt = {inv_meta['time_step_s']*1e6:.4f} µs"
)
print("Loaded from:")
print(" ", fine_dir)
print(" ", inv_dir)

```
